```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns   

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)    

df = pd.read_csv(r"C:\Users\Isabela\Desktop\fraude de cartao\creditcard.csv")     
print("Shape do dataset:", df.shape)
print("\nPrimeiras linhas:")
df.head()

print("=== INFORMAÇÕES DOS DADOS ===")
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")

print("\n=== TIPOS DE DADOS ===")
print(df.dtypes)

print("\n=== VALORES NULOS ===")
print(df.isnull().sum())

print("\n=== ESTATÍSTICAS GERAIS ===")
df.describe()

labels = ['Legítima', 'Fraude']
valores = [284315, 492]
cores = ['steelblue', 'red']

plt.figure(figsize=(8, 5))
sns.barplot(x=labels, y=valores, palette=cores)
plt.title('Distribuição: Legítimas vs Fraudes', fontsize=14)
plt.ylabel('Quantidade')

for i, v in enumerate(valores):
    plt.text(i, v + 1000, f'{v}\n({v/sum(valores)*100:.2f}%)', 
             ha='center', fontsize=11)

plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(x='Class', y='Amount', data=df, palette=['steelblue', 'red'])
plt.xticks([0, 1], ['Legítima', 'Fraude'])
plt.title('Valor das Transações por Classe')
plt.ylabel('Valor (R$)')

plt.subplot(1, 2, 2)
df[df['Class'] == 0]['Amount'].hist(bins=50, color='steelblue', alpha=0.7, label='Legítima')
df[df['Class'] == 1]['Amount'].hist(bins=50, color='red', alpha=0.7, label='Fraude')
plt.title('Distribuição dos Valores')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.legend()

plt.tight_layout()
plt.show()

print("=== VALOR MÉDIO DAS TRANSAÇÕES ===")
print(df.groupby('Class')['Amount'].describe())

plt.figure(figsize=(14, 10))
correlacao = df.corr()
sns.heatmap(correlacao, 
            cmap='coolwarm', 
            center=0,
            fmt='.1f',
            linewidths=0.5)
plt.title('Mapa de Correlação entre Variáveis', fontsize=14)
plt.tight_layout()
plt.show()


print("=== CORRELAÇÃO COM FRAUDE ===")
print(correlacao['Class'].sort_values(ascending=False).head(10))
```
